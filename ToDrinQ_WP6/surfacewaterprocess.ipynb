# ***Surface Water Source***

```python
def surfacewaterprocess(watersource, Partable2):
    # Algae concentration
    print("Is the algae concentration high?:")
    print("1. Yes")
    print("2. No")
    algae_choice = input("Enter your choice (1/2): ")
    algaetreat = "Microsieves" if algae_choice == "1" else None

    # Use of advanced technology (UF or MF)
    print("Use of an advanced technology (UF or MF)?:")
    print("1. Yes")
    print("2. No")
    UF_choice = input("Enter your choice (1/2): ")
    Turbidity = Coag = Floc = flocrem = NOMtreat = OMPtreat = distreat = None
    if UF_choice == "2":
        Turbidity = Partable2['Turbidity'][2]
        print("The average maximum measured turbidity is",Turbidity)
        print("Is turbidity above 1 NTU?:")
        print("1. Yes")
        print("2. No")
        turbin_choice = input("Enter your choice (1/2): ")
        if turbin_choice == "1":
            Coag = "Coagulation"
            Floc = "Flocculation"
            flocrem = "Floc Removal"
            DOC = Partable2['Dissolved Organic Carbon (DOC)'][2]
            print("The average maximum measured DOC is",DOC)
            highNOM = float(input("NOM concentration in the water in (mg/l): "))
            if highNOM > 6:
                NOMtreat = "NOM removal"
        elif turbin_choice == "2":
            print("Is chemical dosing required?")
            print("1. Yes")
            print("2. No")
            coagfloc_choice = input("Enter your choice (1/2): ")
            if coagfloc_choice == "1":
                Coag = "Coagulation"
                Floc = "Flocculation"
            else:
                Coag = Floc = flocrem = NOMtreat = None
    else:
        print("Is OMP removal required?")
        print("1. Yes")
        print("2. No")
        OMP_choice = input("Enter your choice (1/2): ")
        if OMP_choice == "1":
            OMPtreat = "OMP Removal"
        print("Is chemical dosing required?")
        print("1. Yes")
        print("2. No")
        coagfloc_choice = input("Enter your choice (1/2): ")
        if coagfloc_choice == "1":
            Coag = "Coagulation"
        # Hardness
    hardness = Partable2['Hardness'][2]
    print("The average maximum measured hardness is",hardness)
    print("Is hardness in the water high?")
    print("1. Yes")
    print("2. No")
    hard_choice = input("Enter your choice (1/2): ")
    hardtreat = "Softening" if hard_choice == "1" else None

    # Disinfection
    print("Is disinfection required?")
    print("1. Yes")
    print("2. No")
    dis_choice = input("Enter your choice (1/2): ")
    distreat = "Disinfection" if dis_choice == "1" else None

    # Treatment process based on UF
    if UF_choice == '2':
        treatmentprocess = ["Screens", algaetreat, Coag, Floc, flocrem, NOMtreat, hardtreat,
                            "Sand Filtration", "OMP Removal", distreat]
        treatmentprocess = [item for item in treatmentprocess if item is not None]
        treatalt = 0
    else:
        treatmentprocess = ["Screens", algaetreat, Coag, hardtreat,
                            "Ultrafiltration", OMPtreat, distreat]
        treatmentprocess = [item for item in treatmentprocess if item is not None]
        treatalt = 1

    print("Treatment process:", treatmentprocess)

    return treatmentprocess, treatalt
```
