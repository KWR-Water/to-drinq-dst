# ***WTW Flowchart***

```python
import matplotlib.pyplot as plt

def finalflowchart(Treatment):
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.axis('off')

    # Add text annotations
    for ii, tech in enumerate(Treatment["Treatment Technologies"]):
        ax.text(0.1 + 0.05 * ii, 0.9 - 0.07 * ii, tech, color='blue', fontsize=12, ha='center', va='center', bbox=dict(boxstyle='round', facecolor='none', edgecolor='blue'))

    # Add chart title
    ax.set_title("Treatment Plant Flow Chart", fontsize=16, fontweight='bold')


    # Save the figure as an image (optional)
    fig.savefig('flowchart.png', bbox_inches='tight', pad_inches=0.1)

    # Close the figure (optional)
    plt.close(fig)

    return fig
```
