```python
def inletparam(watersource, Partable):
    Partablecolumns = Partable.columns.tolist()
    keypar = []
    Partable2 = None

    if watersource == "Groundwater":
        keypar = ["Temperature", "Dissolved Oxygen (DO)", "Total Alkalinity", "Iron", "Manganese", "Ammonium", "pH", "Arsenic",
                  "Dissolved Organic Carbon (DOC)", "Hardness", "Nitrates", "Total Coliforms"]
    elif watersource == "Surface":
         keypar = ["Temperature", "Turbidity", "Total Coliforms", "Algae", "Escherichia coli", "Iron", "Manganese",
                  "Ammonium", "pH", "Arsenic", "Hardness", "Nitrates", "Lead", "Dissolved Organic Carbon (DOC)"]
    else:
         keypar = ["Temperature", "Turbidity","DO", "TotalAlkalinity", "Total Coliforms","Dissolved Organic Carbon (DOC)", "Algae", "Escherichia coli", "Iron", "Manganese",
                  "Ammonium", "pH", "Arsenic", "Hardness", "Nitrates", "Lead"]

    tf = [col in keypar for col in Partablecolumns]
    Partable2 = Partable.loc[:, tf]

    if "DissolvedOrganicCarbon_DOC_" in Partable2.columns:
        Partable2.rename(columns={"DissolvedOrganicCarbon_DOC_": "DOC"}, inplace=True)

    if "DissolvedOxygen_DO_" in Partable2.columns:
        Partable2.rename(columns={"DissolvedOxygen_DO_": "DO"}, inplace=True)

    return Partable2
```
