# ***Rapid sand filters calculation***


```python
def rapidsandfilter(lanes,units,Partable2,dailyflow,Treatment):
    # Rapid Sand Filter Design
    # This code calculates the dimensions of rapid sand filters in a water treatment plant.
    #treatmenttechnologies2 = treatmenttechnologies.astype(str)

    # Input parameters
    flowRate = max(dailyflow)
    if (Treatment['Treatment Technologies'].str.lower().str.strip() == "single media filtration").any():
        surfaceLoadingRate = 10  # m3/h/m2 average filtration rate for a single media filtration
    elif (Treatment['Treatment Technologies'].str.lower().str.strip() == "dual media filtration").any():
        surfaceLoadingRate = 15  # m3/h/m2 average filtration rate for a dual media filtration
    else:
        surfaceLoadingRate = 8.5  # m3/h/m2 average filtration rate for a moving bed filtration

    # Calculate filter area
    filterArea = flowRate / surfaceLoadingRate

    # Enter the number of lanes

    filterAreaperlane = filterArea / lanes
    filterAreaperunit = filterAreaperlane / units

    # Calculate filter length and width (assuming a rectangular filter)
    filterWidthunit = filterAreaperunit / 4
    filterLengthunit = filterAreaperunit / filterWidthunit

    print("Footprint per unit (m2)",filterAreaperunit)
    print("Footprint of Sand Filtration(m2)",filterAreaperunit*units)
    print("Total number of units",units*lanes)
    print("Total number of lanes",lanes)
    print("Sand filtration Length",filterLengthunit)
    print("Sand filtration width",filterWidthunit*units*lanes)
    return filterArea,filterAreaperlane, filterWidthunit,filterLengthunit
```
