# ***Drinking water treatment plant footprint and energy estimation***


```python
import math as mh
import pandas as pd

import math as mh
import numpy as np
import pandas as pd

def footprint(Treatment, dailyflow, lanes, screen_area_total, Pump_area, Epump, microsieves_area,
                  Coag_area, Ecoag, floc_area, Efloc, clar_area, Eclar, rapidfilterprocessArea,
                  Etot, Eback, ompArea, Eomp, disarea, Edis, stor_area, softarea, softenergy,
                  ufarea, ufenergy, nomenergy, nomarea, aerarea, aerenergy, condarea, condenergy):

    Q = float(dailyflow[0] if isinstance(dailyflow, (list, tuple, np.ndarray)) else dailyflow)
    Q /= 24  # m³/h
    print(f'The nominal flow of the treatment plant is {Q:.2f} m³/h')

    cost_models = {
        "Intake pumping station": {"Area": Pump_area, "Energy": Epump * 365},
        "Screens": {"Area": screen_area_total, "Energy": 0},
        "Microsieves": {"Area": microsieves_area, "Energy": 0},
        "Trickling Filtration": {"Area": Pump_area, "Energy": Epump * 365},
        "Infiltration": {"Area": Pump_area, "Energy": Epump * 365},
        "Coagulation": {"Area": Coag_area, "Energy": Ecoag * 365},
        "Flocculation": {"Area": floc_area, "Energy": Efloc * 365},
        "Floc Removal": {"Area": clar_area, "Energy": Eclar * 365},
        "Sand Filtration": {"Area": rapidfilterprocessArea, "Energy": (Etot + Eback) * 365},
        "OMP Removal": {"Area": ompArea, "Energy": Eomp * 365},
        "NOM removal": {"Area": nomarea, "Energy": nomenergy * 365},
        "Disinfection": {"Area": disarea, "Energy": Edis * 365},
        "Softening": {"Area": softarea, "Energy": softenergy * 365},
        "Conditioning": {"Area": Pump_area, "Energy": condenergy * 365},
        "Ultrafiltration": {"Area": ufarea, "Energy": ufenergy * 365},
        "Storage": {"Area": stor_area, "Energy": 0},
    }

    total_area = 0.0
    total_energy = 0.0

    for _, row in Treatment.iterrows():
        step = str(row["Treatment Steps"]).strip()

        model = cost_models.get(step)
        if model is None:
            print(f"⚠️  No cost model found for: {step}")
            continue

        area = model["Area"]
        energy = model["Energy"]

        print(f"\n🔹 Step: {step}")
        print(f"   - Area: {area:,.2f} m²")
        print(f"   - Energy: {energy:,.2f} kWh/year")

        total_area += area
        total_energy += energy

    print("\n===== Total Summary =====")
    print(f"   Total Area   : {total_area:,.2f} m²")
    print(f"   Total Energy : {total_energy:,.2f} kWh/year")
    print("==========================\n")

    return {"total_energy": total_energy}, {"total_area": total_area}
```
