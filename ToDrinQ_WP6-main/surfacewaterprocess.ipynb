# ***Surface Water Source***

```python
import pandas as pd
import math as mh

def ask_yes_no(question):
    """Helper function for consistent yes/no input."""
    while True:
        print(question)
        print("1. Yes")
        print("2. No")
        choice = input("Enter your choice (1/2): ")
        if choice in ["1", "2"]:
            return choice
        else:
            print("Invalid input. Please enter 1 or 2.")

def surfacewaterprocess(Partable2):
    # Algae concentration
    algae_choice = ask_yes_no("Is the algae concentration high?")
    algaetreat = "Microsieves" if algae_choice == "1" else None

    # Advanced treatment: UF or MF
    UF_choice = ask_yes_no("Is an advanced technology (UF or MF) used?")

    Turbidity = Coag = Floc = flocrem = NOMtreat = OMPtreat = None

    if UF_choice == "2":
        Turbidity = mh. ceil(Partable2['Turbidity'].iloc[2])
        print("The average maximum measured turbidity (NTU) is", Turbidity)
        turbin_choice = ask_yes_no("Is turbidity in the source water high?")
        if turbin_choice == "1":
            Coag = "Coagulation"
            Floc = "Flocculation"
            flocrem = "Floc Removal"
            DOC = mh.ceil(Partable2['Dissolved Organic Carbon (DOC)'].iloc[2]*10)/10
            print("The average maximum measured DOC (mg/l) is", DOC)
            try:
                highNOM = float(input("Enter NOM concentration in the water (mg/l): "))
                if highNOM > 6:
                    NOMtreat = "NOM removal"
            except ValueError:
                print("Invalid input. Skipping NOM treatment decision.")
        else:
            coagfloc_choice = ask_yes_no("Is chemical dosing required?")
            if coagfloc_choice == "1":
                Coag = "Coagulation"
                Floc = "Flocculation"
            else:
                Coag = Floc = flocrem = NOMtreat = None
    else:
        OMP_choice = ask_yes_no("Is OMP removal required?")
        if OMP_choice == "1":
            OMPtreat = "OMP Removal"
        coagfloc_choice = ask_yes_no("Is chemical dosing required?")
        if coagfloc_choice == "1":
            Coag = "Coagulation"
            Floc = "Flocculation"
            flocrem = "Floc Removal"

    # Hardness
    hardness = mh.ceil(Partable2['Hardness'].iloc[2]*10)/10
    print("The average maximum measured hardness (mg/l CaCO3) is", hardness)
    hard_choice = ask_yes_no("Is hardness in the water high?")
    hardtreat = "Softening" if hard_choice == "1" else None

    # Disinfection
    dis_choice = ask_yes_no("Is disinfection required?")
    distreat = "Disinfection" if dis_choice == "1" else None

    # Build treatment process
    if UF_choice == '2':
        treatmentprocess = ["Screens", algaetreat, Coag, Floc, flocrem, NOMtreat, hardtreat,
                            "Sand Filtration", "OMP Removal", distreat]
        treatalt = 0
    else:
        treatmentprocess = ["Screens", algaetreat, Coag, Floc, flocrem, hardtreat,
                            "Ultrafiltration", OMPtreat, distreat]
        treatalt = 1

    treatmentprocess = [step for step in treatmentprocess if step is not None]
    print("Treatment process:", treatmentprocess)

    return treatmentprocess, treatalt
```
