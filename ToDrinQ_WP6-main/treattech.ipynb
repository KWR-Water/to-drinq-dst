# ***Treatment technologies***

```python
def treattech(treatstage, treatmentprocess, treatalt=None):
    def get_user_choice(options, title):
        print(f"{title}:")
        for i, option in enumerate(options, 1):
            print(f"{i}. {option}")
        while True:
            try:
                choice = int(input(f"Enter your choice (1/{len(options)}): "))
                if 1 <= choice <= len(options):
                    return options[choice - 1]
                else:
                    print("Invalid choice. Try again.")
            except ValueError:
                print("Please enter a valid integer.")

    predefined_stages = {
        "Screens": "Screens",
        "Trickling Filtration": "Trickling Filtration",
        "Microsieves": "Microsieves",
        "Infiltration": "Infiltration"
    }

    if treatstage in predefined_stages:
        return predefined_stages[treatstage]

    elif treatstage == "Coagulation":
        return get_user_choice(["Cascade Mixing", "Inline Mixing", "Dynamic Mixing"], "Coagulation")

    elif treatstage == "Flocculation":
        return get_user_choice(["Static mixing", "Dynamic mixing"], "Flocculation")

    elif treatstage == "Floc Removal":
        return get_user_choice(["Sludge Blanket Clarification", "Horizontal Settling", "Titled-plate Settling", "Dissolved Air Flotation"], "Floc Removal")

    elif treatstage == "Sand Filtration":
        return get_user_choice(["Single Media Filtration", "Dual Media Filtration", "Moving Bed Filtration", "Pressurized Filtration Vessel"], "Sand Filtration")

    elif treatstage == "OMP Removal":
        if treatalt == 0:
            options = ["Granular Activated Carbon Filtration (GAC)", "Powdered Activated Carbon Dosing (PAC)", "MP-UV/H2O2", "NF/RO", "Ozonation"]
            tech = get_user_choice(options, "OMP Removal")
            if tech == "Ozonation":
                try:
                    treatmentprocess.remove("Disinfection")
                except ValueError:
                    pass  # Do nothing if not found
            return tech
        else:
            return get_user_choice(["Granular Activated Carbon Filtration (GAC)", "Powdered Activated Carbon Dosing (PAC)", "NF/RO"], "OMP Removal")

    elif treatstage == "NOM removal":
        return get_user_choice(["Anionic IEX", "Enhanced Coagulation"], "NOM removal")

    elif treatstage == "Softening":
        return get_user_choice(["Cationic IEX", "Pellet Softening", "NF/RO"], "Softening")

    elif treatstage == "Aeration":
        return get_user_choice(["Cascades", "Spray Aeration", "Stripping Tower", "Plate Aeration"], "Aeration")

    elif treatstage == "Conditioning":
        return get_user_choice(["Remineralisation", "Dosing of Chemicals", "pH Correction"], "Conditioning")

    elif treatstage == "Disinfection":
        return get_user_choice(["Ozonation", "LP-UV", "Chlorination", "Chloramination"], "Disinfection")

    elif treatstage == "Ultrafiltration":
        try:
            highNOM = float(input("NOM concentration in the water (mg/l): "))
            return "UF" if highNOM > 9.8 else "MF"
        except ValueError:
            print("Invalid NOM concentration.")
            return None

    else:
        print("Error: Invalid treatstage value")
        return None
```
