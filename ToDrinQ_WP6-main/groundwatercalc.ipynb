# ***Groundwater calculations***


```python
def groundwatercalc(lanes,Partable2,dailyflow,Treatment):

  # Check if "Aeration" is in the 'Treatment Steps' column of the DataFrame
    if (Treatment['Treatment Steps'].str.lower().str.strip() == "aeration").any():
        %run "/content/drive/My Drive/Colab Notebooks/WP6/aeration.ipynb"
        aerarea,aerenergy = aeration(lanes,Partable2,dailyflow,Treatment)
        print("")
    else:
        aerarea,aerenergy=0,0

  # Check if "Conditioning" is in the 'Treatment Steps' column of the DataFrame
    if (Treatment['Treatment Steps'].str.lower().str.strip() == "conditioning").any():
        %run "/content/drive/My Drive/Colab Notebooks/WP6/conditioning.ipynb"
        condarea,condenergy,condchem = conditioning(lanes,Partable2,dailyflow,Treatment)
        print("")
    else:
        condarea,condenergy=0,0
        print("")

  # Check if "Sand filtration" is in the 'Treatment Steps' column of the DataFrame
    if (Treatment['Treatment Steps'].str.lower().str.strip() == "sand filtration").any():
        %run "/content/drive/My Drive/Colab Notebooks/WP6/rapidsandfilter.ipynb"
        filterArea,filterAreaperlane, filterWidth,filterLengthunit,Ptot,Pback,Aback, bed_depth, units = rapidsandfilter(lanes,Partable2,dailyflow,Treatment)
        print("")
    else:
        filterArea,filterAreaperlane, filterWidth,filterLengthunit,Ptot,Pback,Aback, bed_depth, units=0,0,0,0,0,0,0,0,0

    # Check if "Disinfection" is in the 'Treatment Steps' column of the DataFrame
    if (Treatment['Treatment Steps'].str.lower().str.strip() == "disinfection").any():
       %run "/content/drive/My Drive/Colab Notebooks/WP6/disinfection.ipynb"
       disarea,disunitarea,disstorage,Pdis,dosageflow,dosageflow2= disin(lanes,Partable2,dailyflow,Treatment)
       print("")
    else:
       disarea,Pdis,dosageflow,dosageflow2=0,0,0,0


   # Check if "Softening" is in the 'Treatment Steps' column of the DataFrame
    if (Treatment['Treatment Steps'].str.lower().str.strip() == "softening").any():
       %run "/content/drive/My Drive/Colab Notebooks/WP6/hardness.ipynb"
       softarea,softenergy= hardness(lanes,Partable2,dailyflow,Treatment)
       print("")
    else:
       softarea, softenergy = 0, 0

   # Check if "Trickling Filtration" is in the 'Treatment Steps' column of the DataFrame
    if (Treatment['Treatment Steps'].str.lower().str.strip() == "trickling filtration").any():
       %run "/content/drive/My Drive/Colab Notebooks/WP6/groundwatfil.ipynb"
       trickarea,trickenergy= trickling(lanes,Partable2,dailyflow,Treatment)
       print("")
    else:
       trickarea,trickenergy = 0, 0


    return aerarea,aerenergy,condarea,condenergy,condchem, filterArea,Ptot,Pback,\
           Aback, disarea,disstorage,Pdis,dosageflow,dosageflow2,softarea, softenergy, trickarea,trickenergy
```
